```sql
use databricks_wanderbricks_dataset_dais_2025.wanderbricks;

show tables;
select * from property_amenities limit 100;
select * from amenities ;
select * from properties limit 10;

create or replace temp view pa_view as 
select 
p.property_id,
a.amenity_id, a.name, a.category,
pr.property_type,
pr.base_price,
c.country
from amenities a
join property_amenities p on a.amenity_id = p.amenity_id
join properties pr on p.property_id = pr.property_id
join cities c on pr.city_id = c.city_id;


select * from pa_view;
-- total amenities

select count(*) from amenities;

-- what amenity category exist.
select distinct(category) from pa_view; 

-- most common amenities
SELECT 
    name AS amenity,
    COUNT(amenity_id) AS usage_count
FROM pa_view
GROUP BY name
ORDER BY usage_count DESC;

-- how many amenities does each property ave
select property_id, COUNT(amenity_id) AS total_amenities
from pa_view
group by property_id
order by total_amenities desc;


-- avg amenities per property
Select round(avg(amenity_count),0) AS avg_amenities_per_property
from (
select property_id, count(distinct amenity_id) AS amenity_count
    from pa_view
    group by property_id
) AS property_amenities_count;

-- most common amenities within each country
select country, name, count(*) as amenity_count, row_number() over(partition by country order by count(*) desc) as rank_no
from pa_view group by country, name 
qualify rank_no
order by country, amenity_count;


--- 7. avg property price by amenity category
select category, round(avg(base_price),1) as avg_price
from pa_view
group by category
order by avg_price desc;
```
